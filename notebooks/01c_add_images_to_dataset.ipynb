```python
import os
import pandas as pd

# Load existing CSV
csv_path = "/content/combined_jewelry_dataset - combined_jewelry_dataset.csv"
df = pd.read_csv(csv_path)

# Folders containing new images
folders = [
    "/content/drive/MyDrive/DATASETS/earring_jhumkas",
    "/content/drive/MyDrive/DATASETS/earrings_randomish"
]

# List to hold new rows
new_data = []

for folder in folders:
    folder_name = os.path.basename(folder)  # get just the folder name
    for file in os.listdir(folder):
        if file.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
            # Only save relative path: folder_name/filename
            image_path = f"{folder_name}/{file}"
            new_data.append({
                "image_path": image_path,
                "description": "",  # you can fill this later
                "jewelry_type": "earring"  # set manually if same type
            })

# Convert to DataFrame
new_df = pd.DataFrame(new_data)

# Append to existing data
updated_df = pd.concat([df, new_df], ignore_index=True)

# Save updated CSV
updated_df.to_csv("updated_data.csv", index=False)

print(f"✅ Added {len(new_data)} new images. Saved as updated_data.csv in Drive")
```
